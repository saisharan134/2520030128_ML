Program 2: Operators

Write a Python program to demonstrate arithmetic operators.
Write a program to demonstrate comparison operators using two numbers.
Write a program to demonstrate logical operators (and, or, not).
Write a program to determine whether a student is eligible for placement based on CGPA and attendance using logical operators.

```python
# ---ARITHMETIC OPERATORS ---
a = 7
b = 2
print("a + b", a + b)
print("a - b", a - b)
print("a * b", a * b)
print("a / b", a / b)
print("a % b", a % b)
print("a ** 2", a ** 2)
print("a // b", a // b)

# --- COMPARISION OPERATIONS ---
cgpa = 8.4
print("cgpa > 8 :", cgpa > 8)
print("cgpa < 8 :", cgpa < 8)
print("cgpa >= 8 :", cgpa >= 8)
print("cgpa < 8 :", cgpa <= 8)
print("cgpa == 8 :", cgpa == 8)
print("cgpa != 8 :", cgpa != 8)

# --- LOGICAL OPERATORS: and / or / not ---
attendance = 85
placed = (cgpa > 85) and (attendance > 80)
print("placed:", placed)
placed = (cgpa > 85) or (attendance > 80)
print("placed:", placed)
placed = not (cgpa > 85)
print("placed:", placed)

```

Program 4: f String

```python
model_name = "Logistic Regression"
accuracy = 0.13408

# Old, clunky way (avoid this):
print("Accuracy of " + model_name + " is " + str(accuracy))

# f-string way (use this):
print(f"Accuracy of {model_name} is {accuracy}")

# ':.3f' inside the braces rounds a float to 3 decimal places -
# you will use this in almost every experiment to print clean metrics.
print(f"Accuracy of {model_name} is {accuracy:.3f}")
```

Program 4: Lists and Indexing

Write a Python program to create a list and access its elements using indexing.
Write a program to display the first, second, and last elements of a list.
Write a program to demonstrate list slicing.
Write a program to add a new element to an existing list using the append() method.
Write a program to display the total number of elements in a list.

```python
# --- LISTS & INDEXING ---

features = ["CGPA", "AptitudeTestScore", "CodingTestScore", "MockInterviewScore"]

print(features)           # the whole list
print(features[0])        # first iten - python counts from 0 not 1
print(features[1])        # second item
print(features[-1])       # last item - -ve index counts from the end
print(len(features))      # how many items are in the list

# Slicing: this gives you sub list
print(features[0:2])

# Adding to a list
features.append("Internships")
print(features)
```

Program 5: Dictionaries

Write a Python program to create a dictionary and display all its key-value pairs.
Write a program to access the value associated with a specific key in a dictionary.
Write a program to iterate through a dictionary using the items() method.
Write a program to store different machine learning models in a dictionary and display their names and objects.

```python
# Create a dictionary
student = {
    "Name": "Sai Sharan",
    "Roll No": 128,
    "CGPA": 9.5
}

# Display all key-value pairs
print("Student Details:")
for key, value in student.items():
    print(key, ":", value)

# Access a specific key
print("\nStudent Name:", student["Name"])

# Machine Learning models dictionary
ml_models = {
    "Linear Regression": "Regression",
    "Decision Tree": "Classification",
    "Random Forest": "Ensemble"
}

print("\nMachine Learning Models:")
for model, model_type in ml_models.items():
    print(model, ":", model_type)
```

Program 6: Conditional Statements

Write a Python program to classify a student's CGPA into High, Mid, or Low categories using if-elif-else.
Write a program to check whether missing values are present in a dataset using conditional statements.
Write a program to determine whether a student is eligible for placement based on CGPA.
Write a program that uses if-elif-else to assign grades based on marks.

```python
# Program 6: Conditional Statements

# 1. Classify student's CGPA
cgpa = 8.3

if cgpa >= 8.5:
    print("Category: High")
elif cgpa >= 6.5:
    print("Category: Mid")
else:
    print("Category: Low")

# 2. Check for missing values in a dataset
data = [10, 20, None, 40]

if None in data:
    print("Missing values are present.")
else:
    print("No missing values.")

# 3. Check placement eligibility
if cgpa >= 7.0:
    print("Student is Eligible for Placement.")
else:
    print("Student is Not Eligible for Placement.")

# 4. Assign grades based on marks
marks = int(input("Enter Marks: "))

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 40:
    print("Grade: D")
else:
    print("Grade: F")
```

Program 7: Loops

Write a Python program to display all elements in a list using a for loop.
Write a program to calculate the average of numbers stored in a list.
Write a program to display only the values greater than a given threshold using a loop.
Write a program to find the maximum and minimum values in a list using loops.
Write a program to count the number of elements satisfying a given condition using a loop.

```python
# Program 7: Loops

numbers = [10, 25, 40, 15, 60, 35, 80]
threshold = 30

# 1. Display all elements
print("List Elements:")
for num in numbers:
    print(num)

# 2. Calculate average
sum = 0
for num in numbers:
    sum += num
average = sum / len(numbers)
print("\nAverage:", average)

# 3. Display values greater than threshold
print("\nValues greater than", threshold, ":")
for num in numbers:
    if num > threshold:
        print(num)

# 4. Find maximum and minimum values
maximum = numbers[0]
minimum = numbers[0]

for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

print("\nMaximum:", maximum)
print("Minimum:", minimum)

# 5. Count elements satisfying a condition
count = 0
for num in numbers:
    if num > threshold:
        count += 1

print("Count of values greater than", threshold, "=", count)
```
